```python
import os

def extend_filename(directory):
    # List all files in the directory
    files = os.listdir(directory)

    # Iterate through each file
    for filename in files:
        # Check if the filename is not a directory
        if not os.path.isdir(os.path.join(directory, filename)):
            # Split the filename and its extension
            name, extension = os.path.splitext(filename)
            
            # If the length of the filename is less than 8 characters, extend it
            if len(name) < 8:
                name = name.ljust(8, 'X')  # Extend to 8 characters with 'X' padding
                
            # Remove any underscores and periods from the filename
            name = name.replace('_', '').replace('.', '')

            # Truncate the filename if it exceeds 8 characters before the dot
            name = name[:8]
            
            # Remove any 'Z' or 'B' from the file extension and replace with 'X'
            extension = extension.replace('Z', 'X').replace('B', 'X')

            # If the length of the extension is less than 4 characters, extend it
            if len(extension) < 4:
                extension = extension.ljust(4, 'X')  # Extend to 3 characters with 'X' padding
                
            # Truncate the extension if it exceeds 3 characters after the dot
            extension = extension[:4]
            
            # Concatenate the extended filename and extension
            new_filename = name[:8] + '.' + extension[1:4]
            
            # Ensure the new filename is unique
            new_filepath = os.path.join(directory, new_filename)
            counter = 1

            while os.path.exists(new_filepath):
                # Adjust the name to include the counter within the 8 character limit
                new_name = name[:2] + 'XXXXXX'[:6-len(str(counter))] + str(counter)
                new_filename = new_name[:8] + '.' + extension[1:4]
                new_filepath = os.path.join(directory, new_filename)
                counter += 1
            
            # Rename the file
            os.rename(os.path.join(directory, filename), new_filepath)

# Replace 'directory_path' with the path of the directory containing the files
directory_path = r'C:\Users\Marble Mpofu\Desktop\Archive error\Cleaned_Files'
extend_filename(directory_path)
print("Done \n------------------------------------------------------------------------------------------------ ")
```
